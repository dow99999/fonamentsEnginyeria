# Resumen Python

## Tipos de datos
- **int**: entero, un numero sin decimales. e.g.: 27
- **float**: decimal, se define con .0. e.g.: 2.45
- **str**: cadena de caracteres, se define con "". e.g.: "hola!"
- **list**: lista modificable de valores, se define con valores entre []. e.g.: [1, "2", 3]
- **tuple**: lista invariable de valores, se define con valores entre (). e.g.: (1, 2, "3")

## Operadores lógicos
Existen los siguientes operadores lógicos con los que podemos evaluar condiciones:
- **not** variable: **negamos** el valor de variable, si su valor era True será False, si era False será True.
- 2 **==** 3: 2 es **igual** a 3, se evalua como falso.
- 2 **!=** 3: 2 es **diferente** de 3, se evalua como cierto.
- 2 **<** 3: 2 es **mas pequeño** que 3, se evalua como cierto.
- 2 **>** 3: 2 es **mas grande** que 3, se evalua como falso.
- 2 **<=** 3: 2 es **mas pequeño o igual** a 3, se evalua como cierto.
- 2 **>=** 3: 2 es **mas grande o igual** a 3, se evalua como falso.

También existen operadores con los que evaluar **múltiples** condiciones a la vez:
- 2 < 3 **and** not variable: para que se evalue como cierto, debe cumplirse que 2 < 3 **y** !variable.
- 2 < 3 **or** 3 == 2: para que se evalue como cierto, debe cumplirse que 2 < 3 **o** 3 == 2.

Podemos juntar tantas condiciones como queramos y mezclar estos dos operadores como queramos. Hay que tener en cuenta que _and_ tiene **prioridad** sobre _or_, o utilizar **paréntesis** en caso de duda:

```python
llueve = True
temperatura = 21
hora = 9

(not llueve and temperatura < 20) or hora < 12
# Que no llueva y que la temperatura sea menor a 20, en caso contrario, que la hora sea menor a 12
#   Con los valores definidos, se evalua como cierto.

```

## Estructuras de selección (condicionales)

### if/elif/else:
Esta estructura sirve para **ejecutar una porción de código** dependiendo del resultado de una **condición evaluada**. Se puede utilizar con un bloque **if**, dos bloques **if+else** o multiples bloques **if+elif+else** (en este podemos poner tantos elif como queramos, incluso podemos omitir el else final).

Podemos definir una **condición a evaluar** tanto en el **if** como en el **elif**, solo si esta condición es cierta entraremos en el bloque de código definido.

```python
if 2 < 3:
    print("2 es mas pequeño que 3, aquí entramos")

elif 1 < 10:
    print("Aquí no entraremos nunca")

else:
    print("Aquí tampoco entraremos")
```

Si la condición de un bloque se evalua como cierta, entraremos en ese bloque y **saldremos** de la estructura if/elif/else, si se evalua como falsa comprobaremos el siguiente bloque de la estructura.

## Estructuras iterativas

### while
El while nos permite ejecutar un trozo de código **mientras** se cumpla una condición. Normalmente se utiliza cuando **no sabemos** cuantas veces se ejecutará un bloque de código.

```python
condicion = True
while condicion:
    print("Me repito")
```

Normalmente, los bucles while tienen dentro alguna pieza de código que modifica la condición evaluada, evitando así un bucle infinito como el anterior:
```python
iterador = 0

while i < 10:
    print("Me repito")
    i += 1
```

### for
El for también nos permite ejecutar un trozo de código **múltiples** veces, pero esas veces **siempre estarán definidas**. Normalmente se utiliza para **recorrer** tipos de datos iterables.
> Algunos de los tipos de datos que hemos visto son iterables: _list_, _tuple_ y _str_. Podemos acceder a una posición en concreto dentro un elemento iterable con `valores[indice]`.

```python
valores = [1,2,3,4,5] # esto es una list

# aquí el for pasa por todos los valores de la list, de izquierda a derecha, y el cuerpo del for muestra por pantalla su valor
for valor in valores:
    print(valor) # muestra por pantalla el valor

```

También se puede definir un valor **iterable** especial con el que hacer N iteraciones, esto se hace con la función range():

```python
# range(10) ejecutará 10 veces el cuerpo del for, asignando a i valores de 0..9
for i in range(10):
    print(i)
```

Se puede obtener la longitud de un elemento iterable con la función len(), de esta forma podemos hacer un recorrido equivalente al del anterior _for_ (el que recorría directamente la _list_) utilizando range:

```python
valores = [1,2,3,4,5]

# Aquí len devuelve 5 y range genera números del 0..4
for i in range(len(valores)):
    print(valores[i]) # Vamos accediendo de izquierda a derecha gracias al índice
```

> Importante: en los valores iterables, el primer índice es el 0, y el último índice es igual a la longitud menos 1.

# Definición de funciones
## Operaciones aritméticas básicas

````python
# OPERACIONES ARITMETICAS BASICAS ###################################################################################################################

def suma(a: float, b: float) -> float:
    """
    Esta función retorna el resultado de la suma dados dos números de tipo float.

    Args:
        a: primer número de la suma
        b: segundo número de la suma

    Returns:
        Retorna el resultado de sumar a + b
    """

    return a + b

def resta(a: float, b: float) -> float:
    """
    Esta función retorna el resultado de la resta dados dos números de tipo float.

    Args:
        a: primer número de la resta
        b: segundo número de la resta

    Returns:
        Retorna el resultado de restar a - b
    """

    return a - b

def multiplicacion(a: float, b: float) -> float:
    """
    Esta función retorna el resultado del producto dados dos números de tipo float.

    Args:
        a: el multiplicando del producto
        b: el multiplicador del producto

    Returns:
        El resultado de multiplicar a * b
    """

    return a * b

def division(a: float, b: float) -> float:
    """
    Esta función retorna el resultado de la división dados dos números de tipo float.

    Args:
        a: el dividendo de la división
        b: el divisor de la división (diferente a 0)

    Returns:
        El resultado de dividir a / b
    """

    return a / b

def resto(a: int, b: int) -> int:
    """
    Esta función retorna el resto de la división entera dados dos números de tipo int.

    Args:
        a: el dividendo de la división
        b: el divisor de la división (diferente a 0)

    Returns:
        El resto de la división entera al dividir a / b
    """

    return a % b

#####################################################################################################################################################
````

## Nota media ponderada

````python
# NOTA MEDIA PONDERADA ##############################################################################################################################

def nota_media_ponderada(notas: list, ponderaciones: list) -> float:
    """
    Esta función retorna el resultado de hacer la media ponderada dadas dos listas,
    una con notas y otra con la ponderación correspondiente de cada nota

    Args:
        notas: lista con las notas del alumnado
        ponderaciones: lista con las ponderaciones

    Returns:
        La nota media ponderada
    """
    
    x = 0
    t_ponderacion = sum(ponderaciones)
    for i in range(len(notas)):
        x += notas[i] * ponderaciones[i]

    return x / t_ponderacion if t_ponderacion != 0 else 0

#####################################################################################################################################################
````

````python
# FUNCIONES ADICIONALES #############################################################################################################################

def potencia_de_2(numero: float) -> float:
    """
    Esta función retorna el resultado elevar 2 a un número decimal cualquiera.

    Args:
        numero: El número al que se queire elevar el 2

    Returns:
        El resultado de elevar 2 al número
    """

    return 2 ** numero


def logaritmo_base_2(numero: float) -> float:
    """
    Esta función retorna el resultado de hacer el logarítmo en base 2 de un número de tipo float.

    Args:
        numero: El numero con el que se quiere hacer el logarítmo (valor superior a 0)

    Returns:
        El resultado de hacer el logarítmo en base 2 del numero
    """

    return math.log2(numero)


def conv_decimal_a_binario(numero: float) -> str:
    """
    Esta función retorna el resultado de convertir un número de tipo float a binario.

    Args:
        numero: El numero que se quiere convertir a binario

    Returns:
        El numero decimal convertido a binario
    """

    return bin(numero)

def conv_binario_a_decimal(numero: str) -> int:
    """
    Esta función retorna el resultado de convertir un número de tipo binario a float.

    Args:
        numero: El numero que se quiere convertir a decimal

    Returns:
        El numero binario convertido a decimal
    """

    return int(numero, 2)

def conversor_factores_unidad(inicio: str, objetivo: str) -> str:
    """
    Esta función toma por parámetro un valor junto a sus unidades y unas unidades objetivo.
    Se espera que las unidades de inicio serán iguales a las objetivo, y que
    lo único que cambiará será su magnitud.

    Args:
        inicio: un numero seguido de sus unidades. e.g.: 27GB
        objetivo: la unidad objetivo a la que hacer la conversión

    Returns:
        El valor convertido a la unidad objetivo junto a su unidad
    """

    UNIDADES = (
        ("m", 0.001),
        ("c", 0.01),
        ("d", 0.1),
        ("", 1),
        ("K", 1000),
        ("M", 1000000),
        ("G", 1000000000),
        ("T", 1000000000000),
        ("P", 1000000000000000),
        ("E", 1000000000000000000)
    )


    def _separar_unidades(valor: str, objetivo: str) -> tuple:
        """
        Esta función separa un valor de sus unidades (una unidad nunca contendrá números). e.g.: 2GB

        Args:
          valor: el valor a separar
          objetivo: la magnitud y unidad objetivo

        Returns:
          El valor en primera posición, su magnitud en segunda y sus unidades en tercera
        """
        num = ""
        unidad = ""
        magnitud = ""

        for c in valor:
            if c in "0123456789":       # parte numérica
                num += c
            elif c == " ":              # los espacios nos los saltamos
                continue
            else:
                unidad += c

        # Giramos las unidades para extraer de derecha a izquierda
        # las unidades y llegar a la magnitud
        magnitud = unidad[::-1]

        # quitamos caracter a caracter las unidades (la magnitud no se extrae, y si se extrae quiere decir que tienen la misma magnitud)
        # Tener la misma magnitud es equivalente a un factor de conversión 1 con la magnitud unidad, asi que no nos preocupa
        for c in objetivo:
            # print("Le quito a", magnitud, c)
            magnitud = magnitud.replace(c, "", 1)

        # le volvemos a dar la vuelta a la magnitud por si acaso, aunque esta debería ser siempre de longitud 1
        magnitud = magnitud[::-1]
        # print("Acabo con magnitud", magnitud)

        # Calculamos la unidad a quitando la magnitud de las unidades extraidas inicialmente (solo en la primera ocurrencia)
        unidad = unidad.replace(magnitud, "", 1)


        # print(num, magnitud, unidad, sep="|")

        # en caso que la magnitud sea igual tanto en el inicio como el objetivo,
        # las unidades tendrán tanto la magnitud como la unidad, y su factor de conversión será 1
        return (float(num), magnitud, unidad)

    def _factor_de_conversion(unidades_i: str, unidades_o: str) -> float|str:
        """
        Esta función calcula el factor de conversión dadas unas magnitudes
        iniciales.

        Args:
          unidades_i: las unidades del valor inicial
          unidades_o: las unidades objetivo

        Returns:
          El factor de conversión entre las dos unidades
        """
        trobats = [None, None]
        i = 0

        # print("fc", unidades_i, unidades_o)

        while ((trobats[0] == None) or (trobats[1] == None)) and i < len(UNIDADES):
            if UNIDADES[i][0] == unidades_i:
                trobats[0] = UNIDADES[i][1]
            if UNIDADES[i][0] == unidades_o:
                trobats[1] = UNIDADES[i][1]

            i += 1

        # Unidades objetivo / unidades iniciales = factor de conversión
        return (trobats[0] / trobats[1]) if ((trobats[0] != None) and (trobats[1] != None)) else "No se puede hacer la conversión"


    n_inicio, m_inicio, u_inicio = _separar_unidades(inicio, objetivo)

    # Le damos la vuelta a objetivo, extraemos las unidades, nos quedamos con la magnitud al revés y le damos la vuelta
    m_objetivo = objetivo[::-1].replace(u_inicio[::-1], "", 1)[::-1]

    # si las magnitudes son las mismas el factor es 1
    fc = _factor_de_conversion(m_inicio, m_objetivo) if m_inicio != m_objetivo else 1

    out = None

    try:
        out = str(n_inicio * float(fc)) + objetivo
    except ValueError:
        out = fc

    # print("Resultado:", out)

    return out


#####################################################################################################################################################
````

# Control de input

````python
def _pedir_valor(msg: str, msge: str, tipo):
    """
    Esta funcion es una ayuda para otras funciones, no está
    pensada para usarse directamente.
    Pide valores hasta que el tipo de valor entrado sea igual al definido por parámetro

    Args:
        msg: el mensaje a mostrar al usuario para pedir el valor
        msge: el mensaje de error a mostrar al usuario en caso de tipo incorrecto
        tipo: la función constructora del valor esperado, e.g: int, float, str...

    Returns:
        El valor con el tipo definido por parámetro

    """

    valor = ""
    correcto = False

    while not correcto:
        try:
            valor = tipo(input(msg)) # Intentamos hacer la conversión al tipo de dato objetivo
            correcto = True
        except ValueError:           # En caso de no poder hacer la conversión, mostramos un msg de error
            print(msge)

    return valor

def pedir_valor_entero(msg: str = "Entra el valor entero: ") -> int:
    """
    Esta funcion pide un valor entero por pantalla y sigue pidiendolo hasta
    que el usuario entre uno válido

    Args:
        msg: [Opcional] El mensaje a mostrar al usuario para pedir el valor

    Returns:
        El valor entero entrado por el usuario
    """

    return _pedir_valor(msg, "Error: el valor debe ser un entero", int)


def pedir_valor_real(msg: str = "Entra el valor real: ") -> float:
    """
    Esta funcion pide un valor real por pantalla y sigue pidiendolo hasta
    que el usuario entre uno válido

    Args:
        msg: [Opcional] El mensaje a mostrar al usuario para pedir el valor

    Returns:
        El valor real entrado por el usuario
    """

    return _pedir_valor(msg, "Error: el valor debe ser un número real", float)
````
