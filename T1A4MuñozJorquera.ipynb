Sumar dos números de la entrada del usuario usando las funciones input(), float() y print().

```python
# Pedimos datos al usuario
num1 = input("Entra un número: ")
num2 = input("Entra otro número: ")

# Cambiamos el tipo de dato de los valores recibidos de str a float
#   de esta forma Python realizará la operación de suma aritmética
#   en vez de una concatenación de strings
num1 = float(num1)
num2 = float(num2)

# Sumamos los valores
resultado = num1 + num2

# Mostramos por pantalla
print(f"El resultado de la suma de {num1} y {num2} es {resultado}")
```

Cambiar de grados Farenheit a Grados Celsius: Para convertir una temperatura de la escala Fahrenheit (°F) a la escala Celsius (°C), se debe restar 32 al valor en Fahrenheit y, posteriormente, dividir el resultado entre 1.8 (o multiplicarlo por 5/9)

Opcional: Solicitar los ºF por teclado utilizando la función input

```python
# Entrada de datos
temp_f = input("Entra una temperatura en Fahrenheit (ºF): ")

# Conversión del valor de str a float para poder realizar
#   operaciones aritméticas
temp_f = float(temp_f)

# Cálculo de la temperatura en Celsius aplicando la fórmula
temp_c = (temp_f - 32) * 5/9

# Mostramos la temperatura convertida
print(f"{temp_f}ºF en grados Celsius es: {temp_c}ºC")
```
