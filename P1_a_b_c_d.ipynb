# Resumen Python

## Tipos de datos
- **int**: entero, un numero sin decimales. e.g.: 27
- **float**: decimal, se define con .0. e.g.: 2.45
- **str**: cadena de caracteres, se define con "". e.g.: "hola!"
- **list**: lista modificable de valores, se define con valores entre []. e.g.: [1, "2", 3]
- **tuple**: lista invariable de valores, se define con valores entre (). e.g.: (1, 2, "3")

## Operadores lógicos
Existen los siguientes operadores lógicos con los que podemos evaluar condiciones:
- **not** variable: **negamos** el valor de variable, si su valor era True será False, si era False será True.
- 2 **==** 3: 2 es **igual** a 3, se evalua como falso.
- 2 **!=** 3: 2 es **diferente** de 3, se evalua como cierto.
- 2 **<** 3: 2 es **mas pequeño** que 3, se evalua como cierto.
- 2 **>** 3: 2 es **mas grande** que 3, se evalua como falso.
- 2 **<=** 3: 2 es **mas pequeño o igual** a 3, se evalua como cierto.
- 2 **>=** 3: 2 es **mas grande o igual** a 3, se evalua como falso.

También existen operadores con los que evaluar **múltiples** condiciones a la vez:
- 2 < 3 **and** not variable: para que se evalue como cierto, debe cumplirse que 2 < 3 **y** !variable.
- 2 < 3 **or** 3 == 2: para que se evalue como cierto, debe cumplirse que 2 < 3 **o** 3 == 2.

Podemos juntar tantas condiciones como queramos y mezclar estos dos operadores como queramos. Hay que tener en cuenta que _and_ tiene **prioridad** sobre _or_, o utilizar **paréntesis** en caso de duda:

```python
llueve = True
temperatura = 21
hora = 9

(not llueve and temperatura < 20) or hora < 12
# Que no llueva y que la temperatura sea menor a 20, en caso contrario, que la hora sea menor a 12
#   Con los valores definidos, se evalua como cierto.

```

## Estructuras de selección (condicionales)

### if/elif/else:
Esta estructura sirve para **ejecutar una porción de código** dependiendo del resultado de una **condición evaluada**. Se puede utilizar con un bloque **if**, dos bloques **if+else** o multiples bloques **if+elif+else** (en este podemos poner tantos elif como queramos, incluso podemos omitir el else final).

Podemos definir una **condición a evaluar** tanto en el **if** como en el **elif**, solo si esta condición es cierta entraremos en el bloque de código definido.

```python
if 2 < 3:
    print("2 es mas pequeño que 3, aquí entramos")

elif 1 < 10:
    print("Aquí no entraremos nunca")

else:
    print("Aquí tampoco entraremos")
```

Si la condición de un bloque se evalua como cierta, entraremos en ese bloque y **saldremos** de la estructura if/elif/else, si se evalua como falsa comprobaremos el siguiente bloque de la estructura.

## Estructuras iterativas

### while
El while nos permite ejecutar un trozo de código **mientras** se cumpla una condición. Normalmente se utiliza cuando **no sabemos** cuantas veces se ejecutará un bloque de código.

```python
condicion = True
while condicion:
    print("Me repito")
```

Normalmente, los bucles while tienen dentro alguna pieza de código que modifica la condición evaluada, evitando así un bucle infinito como el anterior:
```python
iterador = 0

while i < 10:
    print("Me repito")
    i += 1
```

### for
El for también nos permite ejecutar un trozo de código **múltiples** veces, pero esas veces **siempre estarán definidas**. Normalmente se utiliza para **recorrer** tipos de datos iterables.
> Algunos de los tipos de datos que hemos visto son iterables: _list_, _tuple_ y _str_. Podemos acceder a una posición en concreto dentro un elemento iterable con `valores[indice]`.

```python
valores = [1,2,3,4,5] # esto es una list

# aquí el for pasa por todos los valores de la list, de izquierda a derecha, y el cuerpo del for muestra por pantalla su valor
for valor in valores:
    print(valor) # muestra por pantalla el valor

```

También se puede definir un valor **iterable** especial con el que hacer N iteraciones, esto se hace con la función range():

```python
# range(10) ejecutará 10 veces el cuerpo del for, asignando a i valores de 0..9
for i in range(10):
    print(i)
```

Se puede obtener la longitud de un elemento iterable con la función len(), de esta forma podemos hacer un recorrido equivalente al del anterior _for_ (el que recorría directamente la _list_) utilizando range:

```python
valores = [1,2,3,4,5]

# Aquí len devuelve 5 y range genera números del 0..4
for i in range(len(valores)):
    print(valores[i]) # Vamos accediendo de izquierda a derecha gracias al índice
```

> Importante: en los valores iterables, el primer índice es el 0, y el último índice es igual a la longitud menos 1.
